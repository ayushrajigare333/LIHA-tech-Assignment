## About the Dataset

The **international-airline-passengers.csv** dataset contains the total number of international airline passengers (in thousands) for each month from January 1949 to December 1960. It is a classic small time series dataset.

**Source:** Kaggle - International airline passengers (https://www.kaggle.com/datasets/andreazzini/international-airline-passengers).

The dataset contains:

- **144 monthly records (some copies of the file have one extra text line at the bottom, so check the last rows carefully)**
- **2 columns**
- Missing values depend on the copy of the file. You will check and clean them in Section 2

### Columns in the Dataset

| Column | Description |
|---|---|
| Month column | The month of the record (for example `1949-01`) |
| Passengers column | Total passengers in thousands for that month. The original column name is very long, so you will rename it in Section 2 |
