## About the Dataset

The **data.csv** dataset contains measurements calculated from images of breast tumour cells and whether each tumour is malignant or benign.

**Source:** Kaggle - Breast Cancer Wisconsin (Diagnostic) Data Set (https://www.kaggle.com/datasets/uciml/breast-cancer-wisconsin-data).

The dataset contains:

- **569 rows**
- **33 columns (including one extra empty column called `Unnamed: 32`)**
- **The column `Unnamed: 32` contains only missing values**; the other columns are complete

### Columns in the Dataset

| Column | Description |
|---|---|
| `id` | Unique ID of the patient record |
| `diagnosis` | M = malignant, B = benign |
| `radius_mean`, `texture_mean`, `perimeter_mean`, `area_mean`, `smoothness_mean`, `compactness_mean`, `concavity_mean`, `concave points_mean`, `symmetry_mean`, `fractal_dimension_mean` | Average value of 10 measurements of the cell nuclei |
| `radius_se` ... `fractal_dimension_se` | Standard error of the same 10 measurements |
| `radius_worst` ... `fractal_dimension_worst` | Worst (largest) value of the same 10 measurements |
| `Unnamed: 32` | Empty column created by the file format |
