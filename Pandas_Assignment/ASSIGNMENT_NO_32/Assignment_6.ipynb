## Intermediate-Level Questions (21-35)


### Q21: Group the dataset by `track_id` and find the mean of `Unnamed: 0` for each group.


```python
df.groupby('track_id')['Unnamed: 0'].mean()
```

### Q22: Group the dataset by `track_id` and find the sum, mean, and max of `Unnamed: 0`.


```python
df.groupby('track_id')['Unnamed: 0'].agg(['sum', 'mean', 'max'])
```

### Q23: Filter the dataset where `Unnamed: 0` is above the 75th percentile.


```python
percentile_75 = df['Unnamed: 0'].quantile(0.75)
df[df['Unnamed: 0'] > percentile_75]
```

### Q24: Check the total number of duplicated rows in the dataset and drop them if any.


```python
print('Duplicate rows:', df.duplicated().sum())
df.drop_duplicates(inplace=True)
print('Shape after removing duplicates:', df.shape)
```

### Q25: Fill any missing values in the `Unnamed: 0` column with its median.


```python
df['Unnamed: 0'] = df['Unnamed: 0'].fillna(df['Unnamed: 0'].median())
print('Missing values:', df['Unnamed: 0'].isna().sum())
```

### Q26: Drop all rows that have any missing values.


```python
df.dropna(inplace=True)
print('Shape after dropping rows with missing values:', df.shape)
```

### Q27: Convert all string values in `track_id` to uppercase.


```python
df['track_id'] = df['track_id'].str.upper()
df['track_id'].head()
```

### Q28: Find the correlation matrix of all numerical columns in the dataset.


```python
df.select_dtypes(include='number').corr()
```

### Q29: Import matplotlib.pyplot and plot a histogram of `Unnamed: 0`.


```python
plt.figure(figsize=(8, 5))
plt.hist(df['Unnamed: 0'].dropna(), bins=20, edgecolor='black')

plt.show()
```

### Q30: Plot a bar chart showing the top 10 most frequent values in `track_id`.


```python
top_10_track_ids = df['track_id'].value_counts().head(10)
plt.figure(figsize=(12, 6))
top_10_track_ids.plot(kind='bar')

plt.show()
```

### Q31: Plot a scatter plot between `Unnamed: 0` and `popularity`.


```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Unnamed: 0'], df['popularity'], alpha=0.5)

plt.show()
```

### Q32: Group by `artists` and plot the sum of `Unnamed: 0` as a pie chart (top 5 categories).


```python
top_5_artists = df.groupby('artists')['Unnamed: 0'].sum().nlargest(5)
plt.figure(figsize=(8, 8))
top_5_artists.plot(kind='pie', autopct='%1.1f%%', startangle=90)

plt.show()
```

### Q33: Find the row with the maximum `Unnamed: 0` and display its details.


```python
max_index = df['Unnamed: 0'].idxmax()
df.loc[[max_index]]
```

### Q34: Create a pivot table with `track_id` as index and `Unnamed: 0` as values.


```python
pd.pivot_table(df, index='track_id', values='Unnamed: 0', aggfunc='mean')
```

### Q35: Use `.apply()` to create a new column indicating whether `Unnamed: 0` is 'High' (above mean) or 'Low' (below mean).


```python
column_mean = df['Unnamed: 0'].mean()
df['Unnamed: 0_Level'] = df['Unnamed: 0'].apply(lambda value: 'High' if value > column_mean else 'Low')
df[['Unnamed: 0', 'Unnamed: 0_Level']].head()
```
