## About the Dataset

The **WA_Fn-UseC_-Telco-Customer-Churn.csv** dataset contains details of telecom customers, the services they use, their billing information, and whether they left the company (churn).

**Source:** Kaggle - Telco Customer Churn (https://www.kaggle.com/datasets/blastchar/telco-customer-churn).

The dataset contains:

- **7,043 rows**
- **21 columns**
- **`isnull()` shows no missing values at first**, but the `TotalCharges` column has a few blank values stored as spaces

### Columns in the Dataset

| Column | Description |
|---|---|
| `customerID` | Unique ID of the customer |
| `gender` | Male / Female |
| `SeniorCitizen` | 1 if the customer is a senior citizen, otherwise 0 |
| `Partner`, `Dependents` | Whether the customer has a partner / dependents (Yes / No) |
| `tenure` | Number of months the customer has stayed with the company |
| `PhoneService`, `MultipleLines` | Phone service details |
| `InternetService` | DSL, Fiber optic or No |
| `OnlineSecurity`, `OnlineBackup`, `DeviceProtection`, `TechSupport`, `StreamingTV`, `StreamingMovies` | Extra services taken by the customer |
| `Contract` | Month-to-month, One year or Two year |
| `PaperlessBilling` | Whether the customer uses paperless billing |
| `PaymentMethod` | How the customer pays the bill |
| `MonthlyCharges` | Amount charged every month |
| `TotalCharges` | Total amount charged so far |
| `Churn` | Whether the customer left the company (Yes / No) |
