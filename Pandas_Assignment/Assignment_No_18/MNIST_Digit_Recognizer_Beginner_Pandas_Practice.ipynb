## About the Dataset

The **train.csv** dataset contains images of handwritten digits (0 to 9). Each 28 x 28 pixel image is stored as one row of 784 pixel values, along with the digit it shows.

**Source:** Kaggle - Digit Recognizer (https://www.kaggle.com/c/digit-recognizer). Use the `train.csv` file.

The dataset contains:

- **42,000 rows**
- **785 columns (1 label column and 784 pixel columns)**
- **No missing values were found during the initial dataset review**

### Columns in the Dataset

| Column | Description |
|---|---|
| `label` | The digit shown in the image (0 to 9) |
| `pixel0` to `pixel783` | Brightness of each pixel, from 0 (black) to 255 (white) |
