## About the Dataset

The **Tweets.csv** dataset contains tweets (February 2015) about major US airlines, each labelled as `positive`, `neutral` or `negative`, along with the reason for negative tweets.

**Source:** Kaggle - Twitter US Airline Sentiment (https://www.kaggle.com/datasets/crowdflower/twitter-airline-sentiment).

The dataset contains:

- **14,640 rows**
- **15 columns**
- **Missing values are present** in several columns (such as `negativereason`, `negativereason_confidence`, `airline_sentiment_gold`, `negativereason_gold`, `tweet_coord`, `tweet_location` and `user_timezone`)

### Columns in the Dataset

| Column | Description |
|---|---|
| `tweet_id` | Unique ID of the tweet |
| `airline_sentiment` | Sentiment of the tweet: `negative`, `neutral` or `positive` |
| `airline_sentiment_confidence` | How sure the reviewers were about the sentiment (0 to 1) |
| `negativereason` | Reason for a negative tweet (for example `Late Flight` or `Customer Service Issue`) |
| `negativereason_confidence` | How sure the reviewers were about the negative reason |
| `airline` | Airline the tweet is about (American, Delta, Southwest, United, US Airways, Virgin America) |
| `airline_sentiment_gold`, `negativereason_gold` | Extra reviewer labels, almost always empty |
| `name` | Twitter user name |
| `retweet_count` | Number of times the tweet was retweeted |
| `text` | Text of the tweet |
| `tweet_coord` | Location coordinates of the tweet |
| `tweet_created` | Date and time the tweet was posted |
| `tweet_location` | Location written by the user |
| `user_timezone` | Time zone of the user |
