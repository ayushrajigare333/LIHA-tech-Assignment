## About the Dataset

The **StudentPerformance.csv** dataset contains information about **395 students** enrolled in mathematics courses at two Portuguese secondary schools.

It contains **33 columns** describing student demographics, family background, school support, study habits, lifestyle, attendance, and grades.

### Main Column Groups

| Group | Example columns | Description |
|---|---|---|
| Student information | `school`, `sex`, `age`, `address` | Basic demographic and school information |
| Family background | `famsize`, `Pstatus`, `Medu`, `Fedu`, `Mjob`, `Fjob` | Family structure, education, and occupations |
| Study information | `studytime`, `failures`, `schoolsup`, `paid`, `higher` | Study habits and educational support |
| Lifestyle | `freetime`, `goout`, `Dalc`, `Walc`, `health` | Free time, social activity, and health indicators |
| Attendance and grades | `absences`, `G1`, `G2`, `G3` | Absences and first-, second-, and final-period grades |

Grades range from **0 to 20**, and `G3` represents the final grade.

This assignment uses the mathematics-course file from the original dataset. The source uses semicolon-separated values; it has been converted to a standard comma-separated CSV without changing the records.

**Source:** [UCI Machine Learning Repository](https://archive.ics.uci.edu/dataset/320/student+performance)

## 7. Creating and Modifying Columns

### Question 34
Create a new column named **average_grade** containing the row-wise average of `G1`, `G2`, and `G3`.


```python
# df["average_grade"]=df.groupby(["G1","G2","G3"]).mean(axis=1)
df["average_grade"] = df[["G1", "G2", "G3"]].mean(axis=1)
df[["G1","G2","G3","average_grade"]]
```

### Question 35
Create a column named **pass_status** using `apply()` with a `lambda` function. Store `Pass` when `G3` is greater than or equal to 10; otherwise store `Fail`.


```python
df["pass_status"]=df["G3"].apply(lambda x:"Pass" if x>=10 else "Fail")
df['pass_status']
```

### Question 36
