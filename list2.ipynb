```python

#4. Write a Python program to count even and odd numbers in a list.
l1=[5,10,14,20]
ev=0
odd=0

for i in range(0,len(l1)):
    if l1[i]%2==0:
        ev+=1
    else:
        odd+=1

print("even count:",ev)
print("odd count:",odd)
```

```python
#Write a Python program to separate positive and negative numbers from a list.
li=[-5, 3, -2, 7, 0, -1]
pos=[]
neg=[]
for i in li:
    if i>0:
        pos.append(i)
    else:
        neg.append(i)

print("Positive:",pos)            
print("Negative:",neg)            
```

```python
#14. Write a Python program to find common elements between two lists.
l1 = [1, 2, 3, 4]
l2 = [3, 4, 5, 6]
common=[]
for i in l1:
    if i in l2:
        common.append(i)

print("common elements",common)       
```
