```python
#1. Write a Python program to print all natural numbers from 1 to n. - using
#while loop

n=int(input("enter number to print"))
i=1
while i<=n:
    print(i)
    i=i+1

```

```python
#2. Write a Python program to print all natural numbers in reverse (from n to
#1). - using while loop
n=int(input("enter number to print"))
i=n
while i>=1:
    print(i)
    i-=1
    
```

```python
#3. Write a Python program to print all alphabets from a to z. - using while loop

# print(chr(97))65-90 fro A and for 97 to 122 for a
i=97
while i<=122:
    print(chr(i),end='')
    i+=1
```

```python
# 4. Write a Python program to print all even numbers between 1 to 100. -
# using while loop
i=1
while i<=100:
    if i%2==0:
        print(i)
    i+=1
```

```python
#5. Write a Python program to print all odd number between 1 to 100.
i=1
while i<=100:
    if i%2!=0:
        print(i)
    i+=1
```

```python
#6. Write a Python program to find sum of all natural numbers between 1 to n.
n=int(input("enter upto"))
i=1
sum=0
while i<=n:
    sum=sum+i
    i+=1

print(sum)
```
