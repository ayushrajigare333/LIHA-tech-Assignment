## About the Dataset

The **CO2 Emissions_Canada.csv** dataset contains information about different vehicle models and their fuel consumption and CO₂ emission values.

The dataset contains:

### Columns in the Dataset

| Column | Description |
|---|---|
| `Make` | Vehicle manufacturer / brand |
| `Model` | Vehicle model |
| `Vehicle Class` | Category or class of the vehicle |
| `Engine Size(L)` | Engine size in litres |
| `Cylinders` | Number of engine cylinders |
| `Transmission` | Transmission category/code |
| `Fuel Type` | Fuel type category/code |
| `Fuel Consumption City (L/100 km)` | Fuel consumption in city driving |
| `Fuel Consumption Hwy (L/100 km)` | Fuel consumption on highway driving |
| `Fuel Consumption Comb (L/100 km)` | Combined fuel consumption |
| `Fuel Consumption Comb (mpg)` | Combined fuel consumption in miles per gallon |
| `CO2 Emissions(g/km)` | CO₂ emissions in grams per kilometre |
