```python
# 2. Write a Python program to calculate the element-wise sum of two tuples.
t1= (1, 2, 3)
t2= (4, 5, 6)
l=[]
for i in range(len(t1)):
    l.append(t1[i]+t2[i])

print(tuple(l))


```

```python
#3. Write a Python program to find all values that occur more than once in a tuple.
tup=(1, 2, 2, 3, 4, 4, 4)
l=[]
for i in tup:
    if tup.count(i)>1 and i not in l:
        l.append(i)

print("Repeted valuse are",l)        
```
