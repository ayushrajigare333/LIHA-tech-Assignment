```python
#Write a Python program to count the number of vowels and consonants in a given string.
s3="aeiou"
vowels="aeiou"
vowelcount=0
concount=0
for i in s3:
    if i in vowels:
        vowelcount+=1
    else:
        concount+=1

print("the number of vowel is:",vowelcount)
print("the number of consonents is:",concount)
```

```python
st="A@1yush"
uppercnt=0
lowercnt=0
di=0
specialchr=0
for i in range(len(st)):
    if st[i].isupper():
        uppercnt+=1
    elif st[i].islower():
        lowercnt+=1
    elif st[i].isdigit():
        di+=1
    else:
        specialchr+=1            

print("the upper count is :",uppercnt)    
print("the lower case count is :",lowercnt)    
print("the digits count is :",di)    
print("the special character count is :",specialchr)
```
